```python
# Assignment 1: Calculate Undiscounted and Discounted Returns


def main():
    print("=== Reinforcement Learning: Return Calculator ===")

    # 1. Accept number of time steps/episodes
    n_steps = int(input("Enter the number of time steps/episodes: "))

    # 2. Accept reward for each time step
    rewards = []
    print(f"Enter the reward value for each of the {n_steps} time steps:")
    for i in range(n_steps):
        r = float(input(f"  Time step {i + 1} reward: "))
        rewards.append(r)

    # 3. Accept discount factor gamma
    gamma = float(input("Enter the discount factor gamma (0 to 1): "))

    # 4. Calculate Undiscounted Return
    undiscounted_return = sum(rewards)

    # 5. Calculate Discounted Return
    discounted_return = 0.0
    for t, r in enumerate(rewards):
        discounted_return += (gamma**t) * r

    # 6. Display Results
    print("\n" + "=" * 40)
    print("               RESULTS               ")
    print("=" * 40)
    print(f"Entered Rewards Sequence: {rewards}")
    print(f"Discount Factor (gamma) : {gamma}")
    print(f"Undiscounted Return (G0): {undiscounted_return:.4f}")
    print(f"Discounted Return (G0)  : {discounted_return:.4f}")
    print("=" * 40)


if __name__ == "__main__":
    main()
```
