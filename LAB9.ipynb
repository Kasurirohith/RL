```python
gamma = 0.9

# Solve the simultaneous Bellman equations algebraically
V0 = 71.819 / 0.9271
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q0_down = -1 + gamma * V1
Q0_right = -1
Q1_right = 100
Q1_up = -1 + gamma * V0

# Verify state values using the policy probabilities
check_V0 = 0.9 * Q0_down + 0.1 * Q0_right
check_V1 = 0.9 * Q1_right + 0.1 * Q1_up

print("State Values:")
print(f"V(S0) = {V0:.2f}")
print(f"V(S1) = {V1:.2f}")

print("\nAction Values:")
print(f"Q(S0, Down) = {Q0_down:.2f}")
print(f"Q(S0, Right) = {Q0_right:.2f}")
print(f"Q(S1, Right) = {Q1_right:.2f}")
print(f"Q(S1, Up) = {Q1_up:.2f}")

print("\nVerification:")
print(f"Calculated V(S0) = {check_V0:.2f}")
print(f"Calculated V(S1) = {check_V1:.2f}")
```

```python
gamma = 0.9
theta = 0.000001

# Deterministic policy
policy = {
    "S0": "Down",
    "S1": "Right"
}

# Transition model: (next state, probability, reward)
transitions = {
    ("S0", "Down"): [
        ("Wall", 0.1, -1),
        ("S1", 0.9, -1)
    ],
    ("S0", "Right"): [
        ("Wall", 1.0, -1)
    ],
    ("S1", "Right"): [
        ("G", 0.9, 100),
        ("S0", 0.1, -1)
    ],
    ("S1", "Up"): [
        ("S0", 1.0, -1)
    ]
}

# Initialize state values
V = {
    "S0": 0.0,
    "S1": 0.0
}

iterations = 0

# Iterative policy evaluation
while True:
    old = V.copy()

    new_V = {}

    for state in ["S0", "S1"]:
        action = policy[state]
        value = 0.0

        for next_state, prob, reward in transitions[(state, action)]:
            value += prob * (
                reward + gamma * old.get(next_state, 0.0)
            )

        new_V[state] = value

    delta = max(
        abs(new_V["S0"] - old["S0"]),
        abs(new_V["S1"] - old["S1"])
    )

    V = new_V
    iterations += 1

    if delta < theta:
        break

# Calculate action values after convergence
Q = {}

for (state, action), outcomes in transitions.items():
    value = 0.0

    for next_state, prob, reward in outcomes:
        value += prob * (
            reward + gamma * V.get(next_state, 0.0)
        )

    Q[(state, action)] = value

# Display results
print("Number of iterations:", iterations)

print("\nState Values:")
print(f"V(S0) = {V['S0']:.6f}")
print(f"V(S1) = {V['S1']:.6f}")

print("\nAction Values:")
print(f"Q(S0, Down) = {Q[('S0', 'Down')]:.6f}")
print(f"Q(S0, Right) = {Q[('S0', 'Right')]:.6f}")
print(f"Q(S1, Right) = {Q[('S1', 'Right')]:.6f}")
print(f"Q(S1, Up) = {Q[('S1', 'Up')]:.6f}")

print("\nPolicy Consistency Checks:")
print("V(S0) =", f"{V['S0']:.6f}")
print("Q(S0, Down) =", f"{Q[('S0', 'Down')]:.6f}")
print("V(S1) =", f"{V['S1']:.6f}")
print("Q(S1, Right) =", f"{Q[('S1', 'Right')]:.6f}")
```
