```python
import gymnasium as gym
import numpy as np


def run_random_agent(env_id, num_episodes=5):
    """Generic function to run a Random Agent on any Gymnasium environment."""
    print(f"\n==================================================")
    print(f"   RUNNING RANDOM AGENT ON: {env_id}")
    print(f"==================================================")

    # 1. Create Environment
    env = gym.make(env_id)

    total_rewards_all_episodes = []

    # Print action and observation space details
    print(f"Action Space: {env.action_space}")
    print(f"Observation Space: {env.observation_space}\n")

    for ep in range(1, num_episodes + 1):
        # 2 & 3. Reset Environment and Observe Initial State
        state, info = env.reset()
        episode_reward = 0
        step_count = 0
        done = False

        while not done:
            # 4 & 5. Sample a Random Action
            action = env.action_space.sample()

            # 6 & 7. Execute Action in Environment
            next_state, reward, terminated, truncated, info = env.step(action)

            # Accumulate reward
            episode_reward += reward
            step_count += 1

            # 8. Check Episode Termination Criteria
            done = terminated or truncated
            state = next_state

        total_rewards_all_episodes.append(episode_reward)
        print(
            f"Episode {ep:2d} | Total Steps: {step_count:3d} | Total Episode Reward: {episode_reward:6.1f} | Terminated: {terminated} | Truncated: {truncated}"
        )

    # 9. Calculate and Display Overall Statistics
    grand_total_reward = sum(total_rewards_all_episodes)
    avg_reward = grand_total_reward / num_episodes

    print("\n--- Summary Results ---")
    print(f"Total Cumulative Reward ({num_episodes} Episodes): {grand_total_reward:.2f}")
    print(f"Average Reward per Episode: {avg_reward:.2f}")

    env.close()
    return grand_total_reward, avg_reward


if __name__ == "__main__":
    NUM_EPISODES = 5

    # 1. CartPole Environment
    run_random_agent("CartPole-v1", num_episodes=NUM_EPISODES)

    # 2. MountainCar Environment
    run_random_agent("MountainCar-v0", num_episodes=NUM_EPISODES)

    # 3. FrozenLake Environment
    run_random_agent("FrozenLake-v1", num_episodes=NUM_EPISODES)
```
