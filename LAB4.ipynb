```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Part B – Initialize knowledge tables
N = {machine: 0 for machine in machines}       # Participation count
R = {machine: [] for machine in machines}      # Rewards received
Q = {machine: 0.0 for machine in machines}     # Estimated Q-values
UCB = {machine: float('inf') for machine in machines}  # UCB values

total_reward = 0

print("=== Upper Confidence Bound (UCB) Multi-Armed Bandit ===")

# Part C, D and E – UCB selection, reward collection and Q update
for play in range(1, N_play + 1):

    # Calculate UCB values
    for machine in machines:

        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Choose machine with highest UCB
    selected_machine = max(machines, key=lambda x: UCB[x])

    print(f"\nPlay {play}")
    print(f"Selected Machine: {selected_machine}")

    # Get reward from user
    reward = float(input("Enter reward: "))

    # Update participation count
    N[selected_machine] += 1

    # Store reward
    R[selected_machine].append(reward)

    # Incremental sample-average Q-value
    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )

    # Update total reward
    total_reward += reward

    print(f"Reward obtained: {reward}")
    print(f"Updated Q({selected_machine}) = {Q[selected_machine]:.2f}")


# Part F – Display final results
print("\n" + "=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print("\nMachine\tN\tRewards\t\tUCB\t\tQ-value")

# Calculate final UCB values
for machine in machines:
    if N[machine] == 0:
        UCB[machine] = float('inf')
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]}\t"
        f"{UCB[machine]:.2f}\t\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward Received =", total_reward)

# Best machine according to Q-value
best_machine = max(machines, key=lambda x: Q[x])
print("Best Machine =", best_machine)
print("Highest Estimated Q-value =", round(Q[best_machine], 2))
```
